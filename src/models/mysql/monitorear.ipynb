```python
# monitor_bateria.ps1
# Monitorea procesos con alto consumo aproximado
# de recursos que suelen afectar la batería.

while ($true)
{
    Clear-Host

    Write-Host "====================================="
    Write-Host "   MONITOR DE CONSUMO DE RECURSOS"
    Write-Host "====================================="
    Write-Host ""

    # Información de batería
    $battery = Get-CimInstance Win32_Battery

    if ($battery)
    {
        Write-Host "Bateria:"
        Write-Host "Carga restante: $($battery.EstimatedChargeRemaining)%"
        Write-Host "Estado: $($battery.BatteryStatus)"
        Write-Host ""
    }
    else
    {
        Write-Host "No se detecto bateria."
        Write-Host ""
    }

    Write-Host "Procesos con mayor uso de CPU:"
    Write-Host ""

    Get-Process |
    Sort-Object CPU -Descending |
    Select-Object -First 10 `
        Name,
        Id,
        @{
            Name = "CPU_Total"
            Expression = {
                [math]::Round($_.CPU, 2)
            }
        },
        @{
            Name = "RAM_MB"
            Expression = {
                [math]::Round($_.WorkingSet / 1MB, 2)
            }
        } |
    Format-Table -AutoSize

    Write-Host ""
    Write-Host "Actualizando en 5 segundos..."
    
    Start-Sleep 5
}
```
